```python
marketing_data["Education"].value_counts().plot(kind="bar")
plt.title("Customer Distribution by Education")
plt.xlabel("Education")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```

```python
marketing_data["Marital_Status"].value_counts().plot(kind="bar")
plt.title("Customer Distribution by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```

```python
marketing_data["Country"].value_counts().plot(kind="bar")
plt.title("Customer Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```

```python
marketing_data["Income"].plot(kind="hist", bins=20)
plt.title("Distribution of Customer Income")
plt.xlabel("Income")
plt.ylabel("Number of Customers")
plt.show()
```

```python
marketing_data["Recency"].plot(kind="hist", bins=20)
plt.title("Distribution of Customer Recency")
plt.xlabel("Days Since Last Purchase")
plt.ylabel("Number of Customers")
plt.show()
```

```python
marketing_data["Total_Spending"] = (
    marketing_data["MntWines"] +
    marketing_data["MntFruits"] +
    marketing_data["MntMeatProducts"] +
    marketing_data["MntFishProducts"] +
    marketing_data["MntSweetProducts"] +
    marketing_data["MntGoldProds"])
print(marketing_data["Total_Spending"].describe())
```

```python
marketing_data["Total_Spending"].plot(kind="hist", bins=20)
plt.title("Distribution of Total Customer Spending")
plt.xlabel("Total Spending")
plt.ylabel("Number of Customers")
plt.show()
```

```python
channel_data = {
    "Web": marketing_data["NumWebPurchases"].mean(),
    "Catalogue": marketing_data["NumCatalogPurchases"].mean(),
    "Store": marketing_data["NumStorePurchases"].mean()}
print("Average Purchases by Channel:")
print(channel_data)
```

```python
pd.Series(channel_data).plot(kind="bar")
plt.title("Average Purchases by Channel")
plt.xlabel("Purchase Channel")
plt.ylabel("Average Number of Purchases")
plt.xticks(rotation=0)
plt.show()
```

```python
response_counts = marketing_data["Response"].value_counts()
print("Campaign Response:")
print(response_counts)
```

```python
response_counts.plot(kind="bar")
plt.title("Customer Response to Latest Campaign")
plt.xlabel("Response (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.show()
```

```python
campaigns = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5"]
campaign_rates = marketing_data[campaigns].mean() * 100
print("Campaign Acceptance Rates (%):")
print(campaign_rates)
```

```python
campaign_rates.plot(kind="bar")
plt.title("Campaign Acceptance Rates")
plt.xlabel("Campaign")
plt.ylabel("Acceptance Rate (%)")
plt.xticks(rotation=0)
plt.show()
```
